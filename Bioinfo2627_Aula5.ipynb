**1. Exercício 1**

```python
# Adicionar o ficheiro obtido no Clustal Omega à diretoria de trabalho: O nome deve ser Tree1.

# Leitura do ficheiro para um objeto do tipo tree

tree1 = Phylo.read('Tree1', 'newick')

print(tree1)
```

**1.1. Visualização**

```python
# Visualização básica

Phylo.draw_ascii(tree1)
```

```python
# Nomear os nós internos (esta informação não está no ficheiro)

tree1.common_ancestor('Homo', 'Pan').name = 'I1'
tree1.common_ancestor('Rattus', 'Pan').name = 'I2'
tree1.root.name = 'I3'
```

```python
# Repetir a visualização básica (na versão ascii os nós internos não são identificados)

print(tree1)

Phylo.draw_ascii(tree1)
```

```python
# Visualização utilizando Matplotlib

Phylo.draw(tree1)
```

```python
# Visualização utilizando Matplotlib com distâncias nos ramos

fig, ax = plt.subplots(figsize=(8, 4))
Phylo.draw(tree1, axes=ax,
           branch_labels=lambda c: f"{c.branch_length:.3f}" if c.branch_length is not None else None)
```

**1.2. Obtenção de informação de árvores filogenéticas**

```python
# Obter terminais
print('Terminais: ', tree1.get_terminals())

# Obter não terminais
print('Não terminais: ', tree1.get_nonterminals())

# Obter o nó raiz
print('Raiz: ', tree1.root)

```

```python
# Obter distância entre 2 nós

print(tree1.distance({"name": "Rattus"}, {"name": "Pan"}))

print(tree1.distance({"name": "Rattus"}))
```

**1.3. Alteração de árvores filogenéticas**

```python
# Criar cópias da árvore para não se alterar a original

ntA = copy.deepcopy(tree1)

ntB = copy.deepcopy(tree1)
```

```python
# Apagar um nó da árvore

ntA.prune({'name': 'Pan'})

fig, ax = plt.subplots(figsize=(8, 4))
Phylo.draw(ntA, axes=ax,
           branch_labels=lambda c: f"{c.branch_length:.3f}" if c.branch_length is not None else None)
```

```python
# Efetuar a especiação de um nó da árvore (Rattus)

ntB.clade[0, 0].split(2, 0.25)

fig, ax = plt.subplots(figsize=(8, 4))
Phylo.draw(ntB, axes=ax,
           branch_labels=lambda c: f"{c.branch_length:.4f}" if c.branch_length is not None else None)
```

**2. Exercício 2**

```python
# Adicionar o ficheiro obtido no Clustal Omega à diretoria de trabalho: O nome deve ser Tree2.

# Leitura do ficheiro para um objeto do tipo tree

tree2 = Phylo.read('Tree2', 'newick')

print(tree2)
```

```python


#@title Escolha do outgroup
#@markdown Indique o nome **exatamente como aparece no ficheiro Newick**.
#@markdown O outgroup deve ser uma espécie claramente fora do grupo em estudo.
Outgroup = '' #@param {type:"string"}

nomes = [t.name for t in tree2.get_terminals()]
if Outgroup not in nomes:
    raise ValueError(f"'{Outgroup}' não existe na árvore. Nomes válidos: {nomes}")

tree2.root_with_outgroup(Outgroup)

total = sum(c.branch_length for c in tree2.root.clades)
for c in tree2.root.clades:
    c.branch_length = total / 2
```
