**1. Exemplo BLAST**

```python
from Bio import Blast
```

```python
# Obter registo da Uniprot com ID P08567 no formato SeqRec

from Bio import ExPASy, SeqIO

with ExPASy.get_sprot_raw("P08567") as handle:
    seqrec = SeqIO.read(handle, "swiss")
```

```python
# Apresentar alguns detalhes

print(seqrec.id)
print(seqrec.description)
print(seqrec.seq)
```

```python
# Efetuar pesquisa BLAST utilizando a sequência da Uniprot P08567

result_stream = Blast.qblast("blastp", "clustered_nr", seqrec.format("fasta"))
```

```python
# Guardar oresultado num ficheiro xml

with open("P08567_blast.xml", "wb") as out:
    out.write(result_stream.read())
result_stream.close()

with open("P08567_blast.xml", "rb") as f:
    record = Blast.read(f)

print(len(record), "hits")

```

```python
# Visualizar alguns detalhes do primeiro alinhamento

print(record[0])       # resumo do primeiro hit

hit = record[0]
print(len(hit))        # quantos HSPs tem este hit
print(hit[0])          # primeiro HSP (o alinhamento)
```

```python
# Ficheiro 2 - "Exemplo.dnd" contém a árvore filogenética

from Bio import Phylo

tree = Phylo.read('Exemplo.dnd', 'newick')
Phylo.draw_ascii(tree)
```

```python
# Guardar o alinhamento num objeto BioPythyon do tipo multiple sequence alignment

from Bio import AlignIO

msa = AlignIO.read('Exemplo.aln', 'clustal')
```

```python
# Mostrar informação completa

print(msa)
print("Número de sequências:", len(msa))
print("Comprimento do alinhamento:", msa.get_alignment_length())
```

```python
# Mostrar colunas completamente conservadas

cons_colunas = []
for i in range(msa.get_alignment_length()):
    coluna = msa[:, i]
    if coluna.count(coluna[0]) == len(coluna):  # todas iguais
        cons_colunas.append(i)

print("Colunas totalmente conservadas:", cons_colunas)
```
